```python
from geopy.geocoders import Nominatim
from timezonefinder import TimezoneFinder
from datetime import datetime
import pytz
```

```python
def get_local_time(location_name):
    try:
        geolocator = Nominatim(user_agent="clock_up")
        location = geolocator.geocode(location_name)
        if not location:
            return f"Could not find location: {location_name}"

        latitude = location.latitude
        longitude = location.longitude

        tf = TimezoneFinder()
        timezone_str = tf.timezone_at(lat=latitude, lng=longitude)
        if not timezone_str:
            return f"Could not find timezone for: {location_name}"

        timezone = pytz.timezone(timezone_str)
        local_time = datetime.now(timezone)

        time_str = local_time.strftime('%I:%M:%S %p')
        date_str = f"{local_time.month}/{local_time.day}/{local_time.year}"
        return f"The current time in {location_name.title()} is {time_str} and today's date is {date_str}"
    except Exception as e:
        return f"Error: {e}"

# Example usage
location = input("Enter location name: ")
print(get_local_time(location))
```
